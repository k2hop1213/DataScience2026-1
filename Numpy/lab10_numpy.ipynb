```python
# list vs ndarray
import time

a = list(range(100000))
b = list(range(100000,200000))
start = time.time()
for _ in range(1000):
    c = [i + j for i, j in zip(a, b)]
end = time.time()
print(c[:10])
print(f'{end - start:.6f}')
```

```python
na = np.array(a)
nb = np.array(b)
start = time.time()
for _ in range(1000):
    nc = na + nb
end = time.time()

print(nc[:10])
print(f'{end - start:.6f}')
```

```python
def softmax(z):
    z= np.exp(z)
    a =np.sum(z)
    return z/a


logits = np.array([2.0,1.0,-1.0])
probs = softmax(logits)
print(probs)
```

## 2D, 3D arrayes

```python
a = np.random.randint(0,20,size=(3,4,5))
print(a)
print(a.shape)
print(a.ndim)
print(a.mean(axis=(0,2),keepdims=True))
```

```python
# indexing 
a = np.random.randint(0,20,size=(4,5))
print(a)
print(a[2,3])
#b = a[:3,:3]
b = a[::2,::2] # slicing : make a view
print(b)
b[0,0] = 100
print(a)
c = a[:,[0,1]] #fancy indexing : make a copy
print(c)
c[0,0] = 1000
print(a)
```

```python
# transpose
a = np.arange(10).reshape((5,-1))
print(a)
b = a.T
b[0][0] = 100
print(a.T)
```

```python
#broadcasting
a = np.arange(12).reshape((3,-1))
print(a.shape)
# print(a*10)
# b = np.array([1,1,1,1])
b = np.array([1,0,-1])
print(b.shape)
b = b[:,None]
print(b + a)
```

```python
#sum(), mean(), min(), max(), argmin(), argmax()
np.random.seed(1)
a = np.random.randint(0,24,size=(2,3,4))
print(a)
b = a.sum(axis=(0,2))
# a.min()
# a.max()
print(a.mean(axis=(0,2)))
print(b)
print(b.shape)
```

```python
# sorting
a = np.random.randint(1,20,size=(4,5))
print(a)
# a.sort(axis=0) in-place sorting
# print(a)
b = np.sort(a,axis=1) # return a sorted copy
c = np.argsort(a,axis = 1)
print(b)
print(c)
# print(a[c,np.array([i for i in range(c.shape[1])])])
print(a[np.arange(c.shape[0])[:, np.newaxis], c])
print(a[np.arange(c.shape[0])[:, np.newaxis]])
# print(a)
```
