## matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#prepare data
X = np.linspace(0,4*np.pi,100)
Y_sin = np.sin(X)
Y_cos = np.cos(X)
Y_tan = np.tan(X)
Y_tan[np.abs(Y_tan) > 5] = np.nan
# print(X)
```

```python
#Render
fig, ax = plt.subplots()
ax.plot(X,Y_sin,label = 'sin(x)')
ax.plot(X,Y_cos,label = 'cos(x)')
ax.plot(X,Y_tan,label = 'tan(x)')
ax.legend()#각 그래프의 label을 보여주려면 legend사용
ax.set(xlabel='x',ylabel='y')
ax.grid(axis='y')
ax.legend(loc='upper right')
```

```python
def sigmoid(x):
    return 1/(1 + np.exp(-x))

x = np.linspace(-10,10,100)
y = sigmoid(x)
fig, ax = plt.subplots()
ax.plot(x,y,label="sigmoid(x)")
ax.set(xlabel="x",ylabel="y")
ax.set_yticks(np.linspace(0,1,11))
ax.grid(axis='both')
ax.legend()
plt.show()
```

```python
# scatter plot
# histogram
# 2 axes

num_points = 100000
X = np.random.randn(num_points)
Y = X + 0.5*np.random.randn(num_points)

fig , ax = plt.subplots(1,2,figsize=(10,5))
ax[0].scatter(X,Y,edgecolor = 'k',color='r')
ax[0].set(xlabel="X",ylabel="Y")
ax[0].set_title("Scatter plot")

ax[1].hist(X,np.arange(-3,3,0.2),edgecolor='k')
ax[1].set(xlabel="x",ylabel="Frequency")
ax[1].set_title("histogram")

fig.tight_layout()
```

```python
x = np.linspace(-5,5,1000)
y = x**2 + 5

fig,ax = plt.subplots()

ax.plot(x,y,ls=':')
ax.grid(axis = 'both')
ax.set(xlabel='x',ylabel='y')
ax.set_title("f(x)")
plt.show()
# ax.legend()
```

```python
num_points = 100000
X = np.random.uniform(0,10,num_points)
Y = X + 0.5*np.random.uniform(0,10,num_points)

fig , ax = plt.subplots(1,2,figsize=(10,5))
ax[0].scatter(X,Y,edgecolor = 'k',color='r')
ax[0].set(xlabel="X",ylabel="Y")
ax[0].set_title("Scatter plot")

ax[1].hist(X,np.arange(0,11,1),edgecolor='k')
ax[1].set(xlabel="x",ylabel="Frequency")
ax[1].set_title("histogram")
```
