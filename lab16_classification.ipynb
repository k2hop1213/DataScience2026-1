```python
from sklearn import datasets

iris = datasets.load_iris()
```

```python
import numpy as np
X, Y = iris.data, iris.target
print(X.shape,Y.shape)
print(X[:10])
print(iris.feature_names, iris.target_names)
print(Y)
```

```python
# preprocessing
from sklearn.preprocessing import StandardScaler

sc_X = StandardScaler()
X_std = sc_X.fit_transform(X)

print(sc_X.mean_)
print(sc_X.scale_)

print(X_std[:10])

# Note: target class does not need to be standardized
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test,Y_train, Y_test = train_test_split(X_std,Y,test_size=0.3)

print(X_train.shape,X_test.shape)
print(Y_train.shape,Y_test.shape)
```

```python
#model setting

from sklearn import neighbors
from my_knn import MyKNN
# model = neighbors.KNeighborsClassifier(n_neighbors=5)
model = MyKNN(n_neighbors=5)
```

```python
#model fitting
model.fit(X_train,Y_train)
```

```python
# test
Y_pred = model.predict(X_test)
print(Y_test)
print(Y_pred)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score

print(f"confusion matrix: \n {confusion_matrix(Y_test,Y_pred)}")
print(f"accuracy: {accuracy_score(Y_test, Y_pred):.3f}")
print(f"precesion: {precision_score(Y_test,Y_pred,average=None)}")
print(f"recall: {recall_score(Y_test,Y_pred,average=None)}")
print(f"f1: {f1_score(Y_test,Y_pred,average=None)}")
```
