```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
pil_image = Image.open('cat_large.jpeg')
np_image = np.array(pil_image)
```

```python
def better_downscale(image, scale_factor = 10):
    h,w = image.shape[:2]
    h = h//scale_factor
    w = w//scale_factor

    img = image[:h*scale_factor,:w*scale_factor]

    if len(image.shape) == 3: 
        reshaped = img.reshape(h, scale_factor, w, scale_factor, 3)
        downscaled = reshaped.mean(axis=(1, 3))
    else:  
        reshaped = img.reshape(h, scale_factor, w, scale_factor)
        downscaled = reshaped.mean(axis=(1, 3))
        

    return downscaled.astype(np.uint8)





downscaled_image = better_downscale(np_image,10)
plt.imshow(downscaled_image)
```
