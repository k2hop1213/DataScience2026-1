```python
#21
c = [[1,2],
     [8,4],
     [3,9]]
c = np.array(c)
#(a@c)+(b@c) 가 계산이 안되는 이유는 a,b의 열의 수와 c의 행의 개수가 맞지 않기 때문입니다
print((c@a) + (c@b))
```
