## Image Manipulation using Numpy and PIL

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# read an image file
pil_image = Image.open('cat_large.jpeg')
np_image = np.array(pil_image)
```

```python
print(np_image.shape)
print(np_image.dtype)
# print(np_image)
```

```python
# display
# plt.imshow(np_image[100:500,250:750,:]) # crop
# plt.imshow(np_image[:,::-1,:]) #horizontal flip
plt.imshow(np_image[::-1,::-1,:]); #vertical filp
# plt.imshow(np_image[:,:,:]) # 사진 size 줄이기 (scaling) 근데 화질이 안좋아짐, 평균값을 이용
```

```python
pil_image = Image.fromarray(np_image[:,:,:])
pil_image.save("cat_large.jpeg")
```
